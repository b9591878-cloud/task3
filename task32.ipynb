```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x=np.array([1,2,3,4,5])
y=np.array([2,4,6,8,10])
plt.plot(x,y)
plt.show()
```

```python
plt.plot(x,y,"o")
```

```python
x=np.array([127,128,129,130])
y=np.array([100,99,30,10])
plt.plot(x,y)
plt.show()
```

```python
x = np.array([20, 30, 55,32, 60, 70, 90])
plt.plot(x,"o")
plt.show()
```

```python
x=np.array([10,30,50,100])
y=np.array([50,0,30,100])
plt.plot(x,y,marker="o")
plt.show()
```

```python
x=np.array([10,30,50,100])
y=np.array([50,0,30,100])
plt.plot(x,y,marker="x")
plt.show()
```

```python
x=np.array([10,30,50,100])
y=np.array([50,0,30,100])
plt.plot(x,y,marker="X")
plt.show()
```

```python
x=np.array([10,30,50,100])
y=np.array([50,0,30,100])
plt.plot(x,y,"o:g")
plt.show()
```

```python
x=np.array([10,30,50,100])
y=np.array([50,0,30,100])
plt.plot(x,y,marker="o",ms=40,mec="blue",mfc="red") 
plt.show()
```
